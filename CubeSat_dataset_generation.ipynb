```python
import pandas as pd
import random

data = []

for i in range(10000):

    battery_voltage = round(random.uniform(6.0, 8.4), 2)

    battery_current = round(random.uniform(0.5, 3.0), 2)

    temperature = round(random.uniform(-20, 85), 2)

    cpu_usage = round(random.uniform(5, 100), 2)

    memory_usage = round(random.uniform(5, 100), 2)

    signal_strength = round(random.uniform(20, 100), 2)

    solar_panel_power = round(random.uniform(0, 60), 2)

    gyro_x = round(random.uniform(-10, 10), 2)

    gyro_y = round(random.uniform(-10, 10), 2)

    gyro_z = round(random.uniform(-10, 10), 2)

    critical_score = 0
    warning_score = 0

    # Battery Voltage

    if battery_voltage < 6.5:
        critical_score += 1
    elif battery_voltage < 7.0:
        warning_score += 1

    # Battery Current

    if battery_current > 2.8:
        critical_score += 1
    elif battery_current > 2.3:
        warning_score += 1

    # Temperature

    if temperature > 70:
        critical_score += 1
    elif temperature > 55:
        warning_score += 1

    # CPU Usage

    if cpu_usage > 90:
        critical_score += 1
    elif cpu_usage > 75:
        warning_score += 1

    # Memory Usage

    if memory_usage > 90:
        critical_score += 1
    elif memory_usage > 75:
        warning_score += 1

    # Signal Strength

    if signal_strength < 35:
        critical_score += 1
    elif signal_strength < 55:
        warning_score += 1

    # Solar Power

    if solar_panel_power < 5:
        critical_score += 1
    elif solar_panel_power < 15:
        warning_score += 1

    # Gyroscope Stability

    if abs(gyro_x) > 8:
        warning_score += 1

    if abs(gyro_y) > 8:
        warning_score += 1

    if abs(gyro_z) > 8:
        warning_score += 1

    # Final Status

    if critical_score >= 2:
        status = "Critical"

    elif warning_score >= 3:
        status = "Warning"

    else:
        status = "Healthy"

    data.append([
        battery_voltage,
        battery_current,
        temperature,
        cpu_usage,
        memory_usage,
        signal_strength,
        solar_panel_power,
        gyro_x,
        gyro_y,
        gyro_z,
        status
    ])

columns = [
    "BatteryVoltage",
    "BatteryCurrent",
    "Temperature",
    "CPUUsage",
    "MemoryUsage",
    "SignalStrength",
    "SolarPanelPower",
    "GyroX",
    "GyroY",
    "GyroZ",
    "Status"
]

df = pd.DataFrame(data, columns=columns)

df.to_csv("cubesat_telemetry_dataset.csv", index=False)

print("Dataset Generated Successfully")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X= df.drop("Status",axis =1 )
```

```python
y=df["Status"]
```

```python
encoder = LabelEncoder()
```

```python
y_encoded=encoder.fit_transform(y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y_encoded,test_size=0.2,random_state=42)
```

```python
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model Training Complete")
```

```python
accuracy = model.score(X_test, y_test)

print("Accuracy:", accuracy)
```

```python
predictions = model.predict(X_test)

print(predictions[:10])
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        predictions,
        target_names=encoder.classes_
    )
)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, predictions)

print(cm)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions,
    display_labels=encoder.classes_
)

plt.show()
```

```python
import pandas as pd

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print(importance)
```

```python
import matplotlib.pyplot as plt

importance.plot(
    x="Feature",
    y="Importance",
    kind="bar"
)

plt.title("CubeSat Health Feature Importance")
plt.tight_layout()

plt.show()
```
